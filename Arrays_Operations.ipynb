**Array Operations (Arithmatic Operations)**

```
import numpy as np
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])
result = arr1+arr2
print(result)
```

**Broadcasting**

```
l = [10,20,30,40]
arr = np.array(l)
print(arr+10)
```

**Deep Copy**

```
a = np.array([1,2,3,4,5])
b=a.copy()

b[0] = 10
print(a)
print(b)
```

**Shallow Copy**

```
a = np.array([1,2,3,4,5])
b=a.view()

b[0] = 10
print(a)
print(b)
```

**Matrix Operation**

```
A = np.array([[1,2],[4,5]])
B = np.array([[7,8],[10,11]])
print(A)
print(B)
print(A*B)
print(A @ B)
print(np.dot(A,B))
print(A.T)
```
