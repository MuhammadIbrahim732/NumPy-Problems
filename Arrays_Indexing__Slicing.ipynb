**Indexing & Slicing of Vectors**

```python
import numpy as np
arr = np.arange(11,21)
print(arr)
```

```python
print(arr[0])
print(arr[4])
print(arr[:4])
print(arr[1:5:2])
```

**Indexing & Slicing of Matrix**

```python
arr = np.arange(1,31).reshape([6,5])
print(arr)
```

```python
print(arr[0])
print(arr[1,2])
print(arr[0:2,1:3])
print(arr[3:,3:])
print(arr[:,2])
```

**Boolean Indexing**

```python
arr = np.arange(11,21)
print(arr)
```

```python
bool_index = arr%2 == 0
print(bool_index)
```
